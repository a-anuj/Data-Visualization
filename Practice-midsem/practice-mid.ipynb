```python
from scipy.stats import gaussian_kde
import numpy as np
```

```python
data = df["petal_length"]
kde = gaussian_kde(data)

x = np.linspace(data.min(),data.max(),100)

plt.plot(x,kde(x),color="green")
plt.show()
```

```python
x = df["sepal_length"]
y = df["sepal_width"]

xy = np.vstack([x,y])
kde = gaussian_kde(xy)

xi,yi = np.mgrid[x.min():x.max():100j,y.min():y.max():100j]
zi = kde(np.vstack([xi.flatten(),yi.flatten()]))

plt.pcolormesh(xi,yi,zi.reshape(xi.shape),shading="auto",cmap="Blues")
plt.contour(xi,yi,zi.reshape(xi.shape),color="k")

plt.show()
```

```python
from mpl_toolkits.basemap import Basemap
```

```python
m = Basemap(
    llcrnrlat=5,
    llcrnrlon=68,
    urcrnrlat=38,
    urcrnrlon=98,
    projection="merc"
)
m.drawcoastlines()
m.drawrivers(color="lightblue")
m.drawcounties()
m.fillcontinents(color="green",lake_color="blue")
m.drawmapboundary(fill_color='aqua')
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(20,10))

m = Basemap(
    projection="ortho",
    lat_0=20,
    lon_0=80
)


m.drawcoastlines()
m.drawstates()
m.fillcontinents(color="lightgreen",lake_color="aqua")
m.drawmapboundary(fill_color="aqua")

cities = {
    "Delhi" : (28.6,77.2),
    "Cairo" : (30.0,31.2),
    "Moscow" : (55.7,37.6),
    "Bankok" : (13.7,100.5),
    "Nairobi" : (-1.3,36.8)
}

for city,coord in cities.items():
    lat,long = coord
    x,y = m(long,lat)
    plt.plot(x,y,marker="o",color="black")
    plt.text(x,y,city,color="red")

plt.tight_layout()
```

```python
plt.figure(figsize=(20,10))
m = Basemap(
    projection="robin",
    lat_0=0,
    lon_0=0
)
m.drawcoastlines()
m.fillcontinents(color="brown",lake_color="aqua")
m.drawmapboundary(fill_color="aqua")

cities = {
    "New York": (40.7, -74.0),
    "Tokyo": (35.7, 139.7),
    "Sydney": (-33.9, 151.2)
}

for city,coord in cities.items():
    lat,lon = coord
    x,y = m(lon,lat)
    plt.plot(x,y,marker="o",color="green")
    plt.text(x,y,city,color="green")

```

```python
import numpy as np
x = np.linspace(-5,5,100)
y = np.linspace(-5,5,100)

X,Y = np.meshgrid(x,y)

fig = plt.figure(figsize=(10,12))
ax = fig.add_subplot(111,projection="3d")
Z = np.sin(np.sqrt(X**2 + Y**2)) / (np.sqrt(X**2 + Y**2) + 1e-8)

surface = ax.plot_surface(X,Y,Z,cmap="viridis")
```
